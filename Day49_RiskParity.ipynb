```python
# 🚀 DAY 49: RISK PARITY + KELLY
print("🎯 Day48 Top-5 → Vol-Adjusted | Universe Alpha")
print("Production risk parity")

import numpy as np

# Day48 top-5 + vol forecasts
top5 = ['HDFCBANK', 'SBIN', 'TCS', 'LT', 'HCLTECH']
kelly_scores = [0.59, 0.57, 0.56, 0.33, 0.29]
vol_forecasts = [0.031, 0.045, 0.028, 0.038, 0.035]  # Day41 GARCH

# Risk parity: weight ∝ 1/vol * kelly
inverse_vol = 1 / np.array(vol_forecasts)
risk_parity = inverse_vol * np.array(kelly_scores)
risk_parity /= risk_parity.sum()  # Normalize

print("⚖️ RISK PARITY WEIGHTS:")
for i, sym in enumerate(top5):
    print(f"  {sym}: {risk_parity[i]:.0%} (1/{vol_forecasts[i]:.0%}vol * Kelly)")

total_risk = np.sqrt(np.average(vol_forecasts, weights=risk_parity)**2)
print(f"\nPortfolio Vol Target: {total_risk:.1%}")
print(f"Risk-Adjusted Kelly: {total_risk * np.sqrt(0.92):.1%} | Sharpe preserved")

print("\n🎯 DAY 49: Parity production!")
print("• GitHub Day49 → 49/90")
```
