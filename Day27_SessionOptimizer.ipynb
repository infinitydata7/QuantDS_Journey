```python
import pandas as pd
import numpy as np

print("🎯 TIME-BASED ALPHA WINDOWS")
print("Multi-symbol → Hourly regime optimization\n")

# ✅ Fix: lowercase 'h' for pandas 2.2+
hours = pd.date_range("9:15", "15:30", freq="h")

np.random.seed(42)
returns = np.random.randn(len(hours), 3) * 0.005  # 3 assets

# Hourly Sharpe per bar
hourly_sharpe = returns.mean(axis=1) / returns.std(axis=1) * np.sqrt(252 * 6.5)

print("Hourly Sharpe:")
for h, s in zip(hours.time, hourly_sharpe.round(2)):
    print(f"  {h}: {s:+.2f}")

# Top 3 windows
top_hours = np.argsort(hourly_sharpe)[-3:]
print(f"\nTop 3 alpha hours: {[str(hours[i].time()) for i in top_hours]}")

# Session regime labels
session_map = {
    0: "Open Volatility",
    1: "Mid-Morning Drift",
    2: "Pre-Lunch Fade",
    3: "Lunch Slump",
    4: "Afternoon Momentum",
    5: "Power Hour",
    6: "Close Volatility",
}
for i in top_hours:
    label = session_map.get(i, f"Hour-{i}")
    print(f"  ✅ {hours[i].strftime('%H:%M')} → {label} | Sharpe: {hourly_sharpe[i]:.2f}")

print("\n✅ DAY 27: Session optimizer ready!")
print("• GitHub Day27 → 27/90")
```
