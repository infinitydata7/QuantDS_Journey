```python
import vectorbt as vbt
import pandas as pd
import numpy as np

np.random.seed(42)

dates = pd.date_range('2025-01-01', periods=90, freq='D')

prices = pd.DataFrame({
    'HDFCBANK': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod(),
    'SBIN': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod(),
    'TCS': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod()
}, index=dates)

entries = pd.DataFrame({
    'HDFCBANK': np.random.choice([False, True], 90, p=[0.6, 0.4]),
    'SBIN': np.random.choice([False, True], 90, p=[0.55, 0.45]),
    'TCS': np.random.choice([False, True], 90, p=[0.65, 0.35])
}, index=dates)

exits = ~entries

portfolio = vbt.Portfolio.from_signals(
    close=prices,
    entries=entries,
    exits=exits,
    fees=0.0002,
    slippage=0.0005,
    freq='D'
)

print("90-DAY BACKTEST:")
print(f"Total Return: {portfolio.total_return().mean():.1%}")
print(f"Sharpe: {portfolio.sharpe_ratio().mean():.2f}")
print(f"Max DD: {portfolio.max_drawdown().mean():.1%}")
print(f"Win Rate: {portfolio.trades.win_rate().mean():.0%}")
```

```python
print("🎯 Day49 Parity → Full Pipeline | 50/90 Complete")
print("Production validation")

import vectorbt as vbt
import pandas as pd
import numpy as np

np.random.seed(42)

dates = pd.date_range('2025-01-01', periods=90, freq='D')

# Synthetic close prices
close = pd.DataFrame({
    'HDFCBANK': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod(),
    'SBIN': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod(),
    'TCS': 100 * (1 + np.random.normal(0.001, 0.02, 90)).cumprod()
}, index=dates)

# Regime / entry signals
entries = pd.DataFrame({
    'HDFCBANK': np.random.choice([False, True], 90, p=[0.6, 0.4]),
    'SBIN': np.random.choice([False, True], 90, p=[0.55, 0.45]),
    'TCS': np.random.choice([False, True], 90, p=[0.65, 0.35])
}, index=dates)

# Simple exit rule: exit when today's signal turns off after being on
exits = entries.shift(1, fill_value=False) & (~entries)

slippage = 0.0005   # 5 bps
fees = 0.0002       # brokerage

portfolio = vbt.Portfolio.from_signals(
    close=close,
    entries=entries,
    exits=exits,
    fees=fees,
    slippage=slippage,
    freq='D'
)

print("📊 90-DAY BACKTEST:")
print("\nPer asset:")
print((pd.DataFrame({
    'Total Return': portfolio.total_return(),
    'Sharpe': portfolio.sharpe_ratio(),
    'Max DD': portfolio.max_drawdown(),
    'Win Rate': portfolio.trades.closed.win_rate()
}).round(4)))

print("\nAggregate:")
print(f"Total Return: {portfolio.total_return().mean():.1%}")
print(f"Sharpe: {portfolio.sharpe_ratio().mean():.2f}")
print(f"Max DD: {portfolio.max_drawdown().mean():.1%}")
print(f"Win Rate: {portfolio.trades.closed.win_rate().mean():.0%}")

print("\n🎯 DAY 50: Backtest production!")
print("• GitHub Day50 → 50/90 | Phase 4 = 50%")
```
