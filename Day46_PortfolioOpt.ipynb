```python
import numpy as np
from scipy.optimize import minimize

# Day44 live edge + Day41 risk
regime_edge = {'VOL': 0.028, 'DRIFT': 0.015, 'FADE': -0.005}  # Daily alpha
regime_vol = {'VOL': 0.035, 'DRIFT': 0.020, 'FADE': 0.025}
regime_weights = np.array([0.4, 0.45, 0.15])  # Day45 model probs

# Kelly criterion: f = edge / vol^2
kelly_sizes = {}
total_kelly = 0
regime_idx = {'VOL':0,'DRIFT':1,'FADE':2}
for regime in regime_edge:
    edge = regime_edge[regime]
    vol = regime_vol[regime]
    kelly = edge / (vol ** 2)
    kelly_sizes[regime] = max(0, min(1, kelly))  # 0-100%
    total_kelly += kelly_sizes[regime] * regime_weights[regime_idx[regime]]

print("🎯 KELLY OPTIMIZER:")
for regime, size in kelly_sizes.items():
    print(f"  {regime}: {size:.0%} (edge/vol²)")

print(f"\nOptimal Exposure: {total_kelly:.0%}")
print(f"Expected Sharpe: {np.sqrt(total_kelly):.2f}")

# Day41 risk constraint
if total_kelly > 0.5:  # >50% → Half Kelly
    print(f"⚖️ Half-Kelly: Conservative {total_kelly/2:.0%}")

print("\n🎯 DAY 46: Portfolio production!")
print("• GitHub Day46 → 46/90")
```
