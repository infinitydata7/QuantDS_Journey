```python
import numpy as np
import pandas as pd
from scipy.stats import norm

print("🎯 VaR + Stops | Day40 ₹17.8K → Risk Controlled")
print("HFT drawdown protection")

# Position table
positions = pd.DataFrame({
    'symbol': ['HCLTECH', 'KOTAKBANK', 'RELIANCE', 'HDFCBANK'],
    'value_cr': [0.45, 0.54, 0.58, 0.26],   # ₹ Cr
    'vol_pct': [0.035, 0.028, 0.042, 0.031] # Daily vol
})

# Correlation matrix kept separately
corr_matrix = np.array([
    [1.0, 0.4, 0.3, 0.5],
    [0.4, 1.0, 0.2, 0.3],
    [0.3, 0.2, 1.0, 0.4],
    [0.5, 0.3, 0.4, 1.0]
])

portfolio_value = positions['value_cr'].sum()   # ₹ Cr
values = positions['value_cr'].values
vols = positions['vol_pct'].values

# Exposure weights
weights = values / portfolio_value

# Covariance matrix from vol + correlation
cov_matrix = np.outer(vols, vols) * corr_matrix

# Portfolio daily volatility
portfolio_var = weights.T @ cov_matrix @ weights
portfolio_vol = np.sqrt(portfolio_var)

# Daily VaR 95%
z_95 = norm.ppf(0.05)   # inverse CDF / percentile function
var_95_cr = z_95 * portfolio_vol * portfolio_value
var_95_rs = abs(var_95_cr) * 1e7   # 1 Cr = 1e7 rupees

# Max drawdown budget: 5% of portfolio
max_dd_cr = 0.05 * portfolio_value
max_dd_rs = max_dd_cr * 1e7

print("\n🔒 RISK DASHBOARD:")
print(f"Portfolio: ₹{portfolio_value:.2f} Cr")
print(f"Portfolio Daily Vol: {portfolio_vol*100:.2f}%")
print(f"VaR 95% Daily: -₹{var_95_rs:,.0f}")
print(f"Max Drawdown: -₹{max_dd_rs:,.0f}")
print(f"Risk Budget Used: {var_95_rs / max_dd_rs * 100:.1f}%")

# Regime-adaptive stops: 2x daily vol
positions['stop_pct'] = positions['vol_pct'] * 2

print("\n🛑 POSITION STOPS:")
for _, row in positions.iterrows():
    print(f"  {row['symbol']}: {row['vol_pct']*100:.1f}% vol → {row['stop_pct']*100:.1f}% stop")

print("\n🎯 DAY 41: Risk production!")
print("• GitHub Day41 → 41/90")
```
