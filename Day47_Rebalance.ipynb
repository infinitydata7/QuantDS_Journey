```python
# 🚀 DAY 47: LIVE KELLY REBALANCING
print("🎯 Day46 85% Kelly → Auto Orders | Sharpe 0.92")
print("Production portfolio mgmt")

import pandas as pd

# Day46 optimal + Day39 positions
current_portfolio = pd.DataFrame({
    'symbol': ['HCLTECH', 'KOTAKBANK', 'RELIANCE', 'HDFCBANK'],
    'current_weight': [0.25, 0.30, 0.25, 0.20],
    'regime': [0, 1, 1, 0],  # VOL/DRIFT
    'kelly_target': [1.0, 1.0, 1.0, 1.0]  # Day46
})

# Kelly portfolio targets (85% total)
total_kelly = 0.85
current_portfolio['target_weight'] = current_portfolio['kelly_target'] * total_kelly / 3  # Equal regime

# Rebalance trades
current_portfolio['trade_weight'] = current_portfolio['target_weight'] - current_portfolio['current_weight']
current_portfolio['action'] = current_portfolio['trade_weight'].apply(
    lambda x: 'BUY' if x > 0 else 'SELL' if x < 0 else 'HOLD'
)
current_portfolio['trade_size'] = abs(current_portfolio['trade_weight'] * 1.83)  # ₹1.83 Cr

print("🔄 LIVE REBALANCE:")
print(current_portfolio[['symbol', 'action', 'trade_size']].round(2).to_string(index=False))
print(f"\nTarget Exposure: {total_kelly:.0%} | Conservative: 43%")

# Day39 paper execution
for _, row in current_portfolio.iterrows():
    if row['action'] != 'HOLD':
        print(f"📊 PAPER: {row['action']} {row['symbol']} ₹{row['trade_size']:.0f}L")

print("\n🎯 DAY 47: Rebalancing production!")
print("• GitHub Day47 → 47/90")
```
