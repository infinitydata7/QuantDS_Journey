```python
# 🚀 DAY 32: 9-ASSET GPU REGIME PORTFOLIO
print("🎯 REL/HDFC/INFY+ → GPU SIGNAL CONSENSUS")
print("Day31 NIFTY → Multi-Nifty50 parallel regimes")

import torch
import torch.nn as nn
import numpy as np

device = torch.device('cuda')
class RegimeLSTM(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(20, 64, 2, batch_first=True).to(device)
        self.regime_head = nn.Linear(64, 3).to(device)  # Vol/Drift/Fade
    
    def forward(self, x):
        out, _ = self.lstm(x)
        return torch.softmax(self.regime_head(out[:, -1]), dim=-1)

# 9 Nifty50 assets (Day26 portfolio x3)
assets = ['RELIANCE', 'HDFCBANK', 'INFY', 'TCS', 'ICICIBANK', 'HCLTECH', 'SBIN', 'BHARTIARTL', 'KOTAKBANK']
features = np.random.randn(9, 100, 20).astype(np.float32)  # Assets x Ticks x Feats

model = RegimeLSTM()
x = torch.from_numpy(features).to(device)

with torch.no_grad():
    regimes = model(x)  # Batch parallel
    decisions = regimes.argmax(-1).cpu().numpy()
    confidences = regimes.max(-1).values.cpu().numpy()

print("✅ 9-ASSET GPU REGIMES:")
for i, asset in enumerate(assets):
    regime_map = {0:'VOL↑', 1:'DRIFT→', 2:'FADE↓'}
    print(f"  {asset}: {regime_map[decisions[i]]} ({confidences[i]:.1%})")

# Portfolio consensus
portfolio_signal = np.bincount(decisions).argmax()
print(f"\nPortfolio: {'BUY' if portfolio_signal==0 else 'HOLD' if portfolio_signal==1 else 'SELL'}")

print("\n🎯 DAY 32: Multi-asset GPU ready!")
print("• GitHub Day32 → 32/90")
```
