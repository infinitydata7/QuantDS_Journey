```python
# 🚀 DAY 34: GPU REGIME SWITCHING + DYNAMIC SIZING
print("🎯 CONFIDENCE → POSITION SIZE | Day33 Backtest → Live Logic")
print("Regime 40%+ → 2x size | <30% → Cash")

import numpy as np
import pandas as pd

# Regime confidence → Size mapping
confidence_to_size = {
    (0.60, 1.0): 3,   # High conviction
    (0.40, 0.60): 2,  # Medium
    (0.30, 0.40): 1,  # Low
    (0.0, 0.30): 0    # No trade
}

def regime_size(confidence, regime):
    for (low, high), size in confidence_to_size.items():
        if low <= confidence < high:
            return size * (1.5 if regime == 0 else 1.0)  # Vol bonus
    return 0

# 9-asset live simulation
assets = ['RELIANCE', 'HDFCBANK', 'INFY', 'TCS', 'ICICIBANK', 'HCLTECH', 'SBIN', 'BHARTIARTL', 'KOTAKBANK']
confidences = np.random.uniform(0.25, 0.65, 9)  # Day32 range
regimes = np.random.choice([0,1,2], 9)  # Vol/Drift/Fade

print("✅ DYNAMIC SIZING:")
portfolio_size = 0
for i, asset in enumerate(assets):
    size = regime_size(confidences[i], regimes[i])
    regime_name = {0:'VOL↑', 1:'DRIFT→', 2:'FADE↓'}[regimes[i]]
    print(f"  {asset}: {regime_name} {confidences[i]:.0%} → Size {size}")
    portfolio_size += size

print(f"\nPortfolio Exposure: {portfolio_size}/27 max ({portfolio_size/27:.0%})")

print("\n🎯 DAY 34: Regime switching production!")
print("• GitHub Day34 → 34/90")
```
